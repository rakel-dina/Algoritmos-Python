# Análise

- **len()** - Comprimento da frase
- **.count(" ")** - Conta quantas vezes aparece o que colocar entre as aspas nos parênteses | .count(" ",  ,  ), conta com fatiamento, coloca-se números após as vírgulas. 
- **.find(" ")** - Diz em que posição vai encontrar o que está entre as aspas nos parênteses. Se pedir para procurar algo que não esteja na frase, vai retornar o valor '-1'
 - **in** (ex: curso **in** frase) - Vai procurar a palavra curso na variável frase: Retorna True ou False


# Transformação 

- **.replace(" ", " ")** - Troca, reposiciona. Troca a primeira palavra escrita entre aspas pela segunda palavra escrita entre aspas.
- **.upper()** - Deixa todas as letras maiúsculas.
- **.lower()** - Deixa tudo minúsculo.
- **.capitalize()** - As primeiras letras das palavras ficarão maiúsculas.
- **.title()** - Faz o mesmo que o capitalize mas analisa os espaçoes entre as palavras e coloca em maiúsulo a letra que está no início da string.
- **.strip()** - Remove espaço vazios no início e fim da string. Pode usar **rstrip()**, remove espaço à direita e **.lstrip()**, remove espaço à esquerda.
- **.split()** - Ocorre uma divisão considerando os espaços para poder dividir. Gera tecnicamente uma lista.
- **.join()** - Junta a string.

```python
frase = 'Curso em Vídeo Python'
print(frase.count('O'))
print(frase.upper().count("O"))
print(len(frase))
print(len(frase.strip()))
print(frase.replace("Python","Android"))
print('Curso' in frase)
print(frase.find('Vídeo'))
print(frase.find('Video'))
dividido = frase.split()
print(dividido)
print(dividido[2])
print(dividido[2][3])
```
