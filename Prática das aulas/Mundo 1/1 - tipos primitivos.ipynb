 ## Tipos primitivos 

 int = 7 , -2 , 9875 NÚMEROS INTEIROS
 
 float = 4.5, 0.076, -15.223, 7.0 NÚMEROS FLUTUANTES
 
 bool = True, False BOLEANOS
 
 str = 'Olá', '7.5', '' STRINGS

```python
n1 = input('Digite um valor: ') 
print(type(n1))
```

```python
n1 = int(input('Digite um valor: '))
print(type(n1))
```

```python
n1 = int(input('Digite um valor: '))
n2 = int(input('Digite um valor: '))
s = n1 + n2
print(f'A soma vale: {s}')
```

**Retirando o int, vire uma uma concatenação porque o tipo primitivo passa a ser 'str'**

```python
n1 = (input('Digite um valor: '))
n2 = (input('Digite um valor: '))
s = n1 + n2
print(f'A soma vale: {s}')
```

```python
n = input("Digite algo algo: ")
print('')
print(f"{n} é númérico? {n.isnumeric()}")
print('---------------')
print(f"{n} é alpha? {n.isalpha()}")
print('---------------')
print(f"{n} é alphanumérico? {n.isalnum()}")
```

```python
n = input("Digite algo algo: ")
print(f'"{n}" está em letras maiúsculas? {n.isupper()}') 
```
