# Utilizando Módulos  

### math 
- ceil - faz arredondamento pra cima
- floor - faz arredondamento pra baixo
- trunc - alimina da vírgula pra frente sem realizar o arredondamento 
- pow -potência
- sqrt - para calcular a raiz quadrada
 - factorial - calcular fatorial

import math - importa tudo <p>
from math import sqrt - para importa apenas um 

```python
import math 
num = int(input('Digite um número: '))
raiz = math.sqrt(num)
print(f'A raiz de num é {raiz}')
```

```python
from math import sqrt, floor
num = int(input('Digite um número: '))
raiz = sqrt(num)
print(f'A raiz de num é {floor(raiz)}') # arredonda pra baixo
```
