# Cores no terminal

**\033[ m** : Entre o colchete e a letra 'm' será escrito o cógido do 'style', 'text', 'back'

**style** | **text** | **back**
---|---|---
0 None (nenhum) | 30 (Branco) | 40 (Branco)
1 Bold (Negrito) | 31 (Vermelho) | 41 (Vermelho)
4 Underline (Sublinhado)| 32 (Verde) | 42 (Verde)
7 Negative (Negativo) | 33 (Amarelo) | 43 (Amarelo)
---| 34 (Azul) | 44 (Azul)
---| 35 (Roxo) | 45 (Roxo)
---| 36 (Ciano) | 46 (Ciano)
---| 37 (Cinza) | 47 (Cinza)


```python
print("\033[31mOlá, mundo!\033[m")
print("\033[7;31;43mOlá, mundo!")
```

```python
a = 3
b = 5 
print(f"Os valores são \033[33m{a}\033[m e \033[34m{b}!!!")
```

```python
nome = 'Raquel'
cores = {'limpa':'\033[m',
         'azul':'\033[34m',
         'amarelo':'\033[33m',
         'pretobranco':'\033[7:30m]'}
print(f"Olá {cores['azul']}{nome}!!!")
```
