## Estrutura de repetição while

```python
c = 1
while c < 10:
    print(c)
    c = c +1
print('Fim')
```

```python
n = 1
while n != 0: # flag - condição de parada
    n = int(input('Digite um número | [0] encerra: '))
 
print('FIM')
```

```python
r = 'S'
while r == 'S':
    n = int(input('Digite um valor: '))
    r = str(input("Quer continuar? [S] ")).upper()
print('FIM')
```

```python
n = 1
par = impar = 0
while n != 0:
    n = int(input("Digite um número: "))
    if n != 0:
        if n % 2 == 0:
            par = par + 1
        else:
            impar = impar + 1
print(f"Foram digitados {par} números pares e {impar} números ímpares.")
```
