**57 - Faça um programa que leia o sexo de uma pessoa, mas só aceite os valores ‘M’ ou ‘F’. Caso esteja errado, peça a digitação novamente até ter um valor correto.**

```python
sexo = str(input("Qual é o sexo? [F] [M]: ")).strip().upper()
while sexo not in 'FM':
    print(f'INVÁLIDO. Digite apenas [F] ou [M].')
    sexo = str(input("Qual é o sexo? [F] [M]: ")).strip().upper()
print(f'Sexo [{sexo}] registrado com sucesso.')
```

**58 - Melhore o jogo do DESAFIO 28 onde o computador vai “pensar” em um número entre 0 e 10. Só que agora o jogador vai tentar adivinhar até acertar,
mostrando no final quantos palpites foram necessários para vencer.**

```python
from random import randint
computador = randint(0,10)
print(computador)
print('[ADIVINHE UM NÚMERO DE 0 A 10 QUE O COMPUTADOR PENSOU]')
certo = False
while not certo:
    jogador = int(input('Digite: '))
    if jogador == computador: 
        certo = True
print('PARABÉNS! ACERTOU! ')    
```

```python
computador = randint(0,10)
#print(computador)
palpites = 0
print('[ADIVINHE UM NÚMERO DE 0 A 10 QUE O COMPUTADOR PENSOU]')
jogador = int(input('Digite: '))
while jogador != computador:
    print("ERRADO, Tente Novamente: ")
    jogador = int(input('Digite: '))
    palpites = palpites + 1
    if jogador == computador: 
        print(f'PARABÉNS! Voce acertou com {palpites} palpites.!')
print("Fim")
```
