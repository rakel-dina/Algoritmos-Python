# Estrutura de repetição for
### Laço com variável de controle 
**for c in range(1,10):** (Fazer o contador no intervalo de 1 a 10)

```python
for c in range(1,7):
    print(f"{c} - Oi")
print("Fim")
```

```python
for c in range(7,0,-1):
    print(f"{c} - Oi")
print("Fim")
```

```python
for c in range(0,7,2):
    print(f"{c} - Oi")
print("Fim")
```

```python
n = int(input("Digite um número: "))
for c in range(0, n+1):
    print(c)
print("FIM")
```

```python
i = int(input("Início: "))
f = int(input("Fim: "))
p = int(input("Passo: "))
for c in range(i, f+1, p):
    print(c)
```

```python
s = 0
for c in range(0,3):
    n = int(input("Digite um número: "))
    s = s + n
print(f"O somatório dos três números é {s}")
print("FIM")
```
